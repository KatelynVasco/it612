```python
%%writefile code.js
// Part A
// Loop that checks if a number is less than 10 and doubles it
const data = [5, 12, 3, 20, 11, 9, 30];

let result1 = [];
for (let number of data) {
  if (number > 10) {
    result1.push(number * 2);
  }
}
console.log(result1);

// Pipeline that checks if a number is less than 10 and doubles it
let result2 = data.filter(x => x > 10).map(x => x * 2);
console.log(result2);

// Part B
// Loop to compute sum of doubled evens
const numbers = [1, 2, 3, 4, 5, 6];
let sum = 0
for (let x of numbers) {
  if (x % 2 === 0) {
    sum += (x *2);
  }
}
console.log(sum);

// Pipeline to compute sum of doubled evens
let sum2 = numbers.filter(x => x % 2 === 0).map(x => x * 2).reduce((acc, x) => acc + x);
console.log(sum2);

// Part C
// some, find, and forEach methods
const users = [
  { id: 1, name: "Alice", age: 25 },
  { id: 2, name: "Bob", age: 17 },
  { id: 3, name: "Carol", age: 19 }
];

console.log(users.some(x => x.age >= 18))
console.log(users.find(x => x.id === 3))
users.forEach(x => console.log(x.name))

// Part D
// Higher-Order Function
function repeat(n, action) {
  for (let i = 0; i < n; i++) {
    action(i);
  }
}
repeat(3, console.log);
repeat(4, i => console.log("Step " + i));
```
